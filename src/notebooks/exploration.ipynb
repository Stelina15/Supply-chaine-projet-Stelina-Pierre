```python
import networkx as nx

G = nx.DiGraph()

edges = [
    ("E", "a", 5),
    ("E", "b", 10),
    ("E", "e", 8),
    ("a", "c", 7),
    ("a", "d", 10),
    ("b", "c", 8),
    ("b", "d", 2),
    ("b", "e", 1),
    ("c", "g", 7),
    ("d", "g", 4),
    ("d", "S", 6),
    ("d", "f", 2),
    ("e", "f", 4),
    ("f", "S", 6),
    ("g", "S", 10),
]

for u, v, cap in edges:
    G.add_edge(u, v, capacity=cap)

flow, _ = nx.maximum_flow(G, "E", "S")
print("Flot sans contraintes :", flow)
```

```python
G2 = nx.DiGraph()

capacites_villes = {
    "a": 6,
    "b": 7,
    "c": 8,
    "d": 6,
    "e": 6,
    "f": 5,
    "g": 9,
}
```

```python
for ville, cap in capacites_villes.items():
    G2.add_edge(f"{ville}_in", f"{ville}_out", capacity=cap)

for u, v, cap in edges:
    u2 = f"{u}_out" if u in capacites_villes else u
    v2 = f"{v}_in" if v in capacites_villes else v
    G2.add_edge(u2, v2, capacity=cap)

flow2, _ = nx.maximum_flow(G2, "E", "S")
print("Flot avec contraintes villes :", flow2)
```
